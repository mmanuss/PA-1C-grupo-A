```python
# Escrever um programa que o usúario digita um número e o programa imprimi de 0 até o número;

numero = int(input("digite um número: "))
for i in range(numero + 1):
    print(i)
```

```python
# Escrever um programa que o usúario digita um número e o programa imprimi em ordem descrescente
# até o zero;

numero = int(input("digite um número: "))
for i in range (numero + 1):
   print (numero- i)
```

```python
# Peça um número e mostre todos os números pares até o número;

numero = int(input("digite um número: "))
for i in range (0,numero + 1, 2):
   print(i)
```

```python
# Peça um número e calcule a soma de 1 até o número;

numero = int(input("Digite um número para somar: "))
soma = 0
for i in range(1, numero + 1):
    soma += i
print(f"A soma de 1 até {numero} é: {soma}")
```

```python
# Peça uma palavra e conte as letras: Ex.; Python - A palavra tem 06 letra

palavra = input("Digite uma palavra: ")
quantidade_letras = len(palavra)
print(f"A palavra '{palavra}' tem {quantidade_letras} letras.")
```

```python
# Peça uma palavra e conte as vogais;

palavra = input("Digite uma palavra: ")
vogais_set = "aeiou"
quantidade_vogais = 0

for letra in palavra.lower():
    if letra in vogais_set:
        quantidade_vogais += 1

print(f"A palavra '{palavra}' tem {quantidade_vogais} vogais.")
```

```python
# Peça o usúario para digitar cinco números e diga qual é maior

numeros = []
for i in range(5):
    num = int(input(f"Digite o {i+1}º número: "))
    numeros.append(num)

maior_numero = max(numeros)
print(f"O maior número digitado é: {maior_numero}")
```
