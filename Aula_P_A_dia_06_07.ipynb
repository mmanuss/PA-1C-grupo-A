```python
#2 – Contador de Dígitos
numero = input("Digite um número inteiro positivo: ")


quantidade_digitos = len(numero)

print(f"Quantidade de dígitos: {quantidade_digitos}")
```

```python
#3 – Cofrinho
meta = 10000.0
total_acumulado = 0.0
mes = 0

while total_acumulado < meta:
    mes += 1
    deposito = float(input(f"Informe o valor do depósito para o mês {mes}: R$ "))
    total_acumulado += deposito

print("\n--- Meta Atingida! ---")
print(f"Quantos meses foram necessários: {mes}")
print(f"Total acumulado: R$ {total_acumulado:.2f}")
print(f"Em qual mês a meta foi atingida: Mês {mes}")
```

```python
#4 – Caixa Registradora

total_da_compra = 0.0

while True:
    preco = float(input("Preço do produto (ou 0 para encerrar): R$ "))
    if preco == 0:
        break

    quantidade = int(input("Quantidade comprada: "))

    subtotal = preco * quantidade


    if quantidade < 5:
        porcentagem_desconto = 0
    elif quantidade < 10:
        porcentagem_desconto = 5
    elif quantity < 20:
        porcentagem_desconto = 10
    else:
        porcentagem_desconto = 15

    desconto = subtotal * (porcentagem_desconto / 100)
    valor_final = subtotal - descuento

    print(f"Subtotal: R$ {subtotal:.2f}")
    print(f"Desconto ({porcentagem_desconto}%): R$ {desconto:.2f}")
    print(f"Valor final do produto: R$ {valor_final:.2f}")
    print("-" * 30)

    total_da_compra += valor_final

print(f"\nCompra encerrada. Total geral a pagar: R$ {total_da_compra:.2f}")
```
