```python
# 3 - Programa para calcular a soma de números digitados pelo usuário
soma_total = 0
print("Digite os números para somar. Pressione Enter sem digitar nada para finalizar.")

while True:
    entrada_usuario = input("Digite um número (ou Enter para finalizar): ")
    if not entrada_usuario:
        break
    try:
        numero = float(entrada_usuario)
        soma_total += numero
    except ValueError:
        print("Entrada inválida. Por favor, digite um número válido.")

print(f"A soma dos números digitados é: {soma_total}")
```

```python
# 8 - Programa que recebe valores e quantidades de produtos e exibe o produto com o maior valor total

maior_valor_total = 0
nome_produto_maior = ""

print("Vamos registrar produtos e suas quantidades para encontrar o que tem o maior valor total.")
print("Digite 'fim' a qualquer momento no nome do produto para terminar.")

while True:
    nome_produto = input("Digite o nome do produto (ou 'fim' para finalizar): ").strip()
    if nome_produto.lower() == 'fim':
        break
    if not nome_produto:
        print("Nome do produto não pode ser vazio. Tente novamente.")
        continue

    try:
        preco_unitario = float(input(f"Digite o preço unitário de '{nome_produto}': "))
        if preco_unitario < 0:
            print("O preço não pode ser negativo. Tente novamente.")
            continue

        quantidade = int(input(f"Digite a quantidade de '{nome_produto}': "))
        if quantidade < 0:
            print("A quantidade não pode ser negativa. Tente novamente.")
            continue

        valor_total_produto = preco_unitario * quantidade

        if valor_total_produto > maior_valor_total:
            maior_valor_total = valor_total_produto
            nome_produto_maior = nome_produto

    except ValueError:
        print("Entrada inválida. Por favor, digite um número para o preço ou quantidade.")
    except Exception as e:
        print(f"Ocorreu um erro inesperado: {e}")

if nome_produto_maior:
    print(f"\nO produto com o maior valor total é '{nome_produto_maior}' com um valor total de R$ {maior_valor_total:.2f}.")
else:
    print("Nenhum produto foi registrado.")
```

```python
# 9 - Programa que pede ao usuário para digitar 5 números e exibe o maior número

numeros = []

print("Por favor, digite 5 números:")

for i in range(1, 6):
    while True:
        try:
            numero_digitado = float(input(f"Digite o {i}º número: "))
            numeros.append(numero_digitado)
            break
        except ValueError:
            print("Entrada inválida. Por favor, digite um número válido.")

if numeros:
    maior_numero = max(numeros)
    print(f"\nOs números digitados foram: {numeros}")
    print(f"O maior número digitado é: {maior_numero}")
else:
    print("Nenhum número foi digitado.")
```

```python
# 10 - Jogo de Adivinhação

import random

numero_secreto = random.randint(1, 10)
tentativas = 0
max_tentativas = 3

print("Bem-vindo ao Jogo de Adivinhação!")
print(f"Estou pensando em um número entre 1 e 10. Você tem {max_tentativas} tentativas.")

while tentativas < max_tentativas:
    try:
        chute = int(input(f"Tentativa {tentativas + 1}/{max_tentativas}: Digite seu chute: "))

        if chute < 1 or chute > 10:
            print("Por favor, chute um número entre 1 e 10.")
            continue

        if chute == numero_secreto:
            print(f"Parabéns! Você acertou o número secreto ({numero_secreto}) em {tentativas + 1} tentativas!")
            break
        elif chute < numero_secreto:
            print("O número secreto é MAIOR que o seu chute.")
        else:
            print("O número secreto é MENOR que o seu chute.")

        tentativas += 1

    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")

if tentativas == max_tentativas and chute != numero_secreto:
    print(f"Suas {max_tentativas} tentativas acabaram. O número secreto era {numero_secreto}.")
print("Fim do jogo.")
```
