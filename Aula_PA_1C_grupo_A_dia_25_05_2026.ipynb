```python
#Escrever um programa que recebe 4 notas, calcula a média e imprimi se o aluno foi aprovado, recuperação ou reprovado

#nota maior ou igual a 7 = aprovado
#nota maior ou igual a 5 = recuperação
#nota menor que 5 = reprovado

# Inicializa uma lista para armazenar as notas
notas = []

# Loop para receber as 4 notas do usuário
for i in range(4):
    while True:
        try:
            nota = float(input(f"Digite a {i+1}ª nota: "))
            if 0 <= nota <= 10: # Valida se a nota está entre 0 e 10
                notas.append(nota)
                break
            else:
                print("Nota inválida. Por favor, digite um valor entre 0 e 10.")
        except ValueError:
            print("Entrada inválida. Por favor, digite um número.")

# Calcula a média das notas
media = sum(notas) / len(notas)

print(f"\nAs notas digitadas foram: {notas}")
print(f"A média do aluno é: {media:.2f}") # Exibe a média com 2 casas decimais

# Verifica a situação do aluno com base na média
if media >= 7:
    print("Situação: Aprovado!")
elif media >= 5:
    print("Situação: Recuperação.")
else:
    print("Situação: Reprovado.")
```
